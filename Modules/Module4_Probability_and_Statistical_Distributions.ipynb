## 1. Fundamentals of Probability

Probability measures how likely an event is to occur, expressed as a number between 0 (impossible) and 1 (certain).

Key terms:
- **Experiment**: a process with an uncertain outcome (e.g. rolling a die)
- **Sample space**: the set of all possible outcomes
- **Event**: a subset of the sample space
- **P(A)**: probability of event A occurring


```r
# Simple probability calculation: probability of rolling a 4 on a fair die
sample_space <- 1:6
event <- 4
prob <- length(event) / length(sample_space)
prob

# Simulating a die roll many times to approximate the true probability
set.seed(123)
rolls <- sample(1:6, size = 10000, replace = TRUE)
mean(rolls == 4)   # approximate probability of rolling a 4
```

## 2. Probability Rules and Applications

- **Addition rule**: P(A or B) = P(A) + P(B) - P(A and B)
- **Multiplication rule**: P(A and B) = P(A) x P(B) if independent
- **Complement rule**: P(not A) = 1 - P(A)
- **Conditional probability**: P(A given B) = P(A and B) / P(B)


```r
# Example: drawing cards from a standard 52 card deck
p_heart <- 13/52
p_king <- 4/52
p_king_of_hearts <- 1/52

# Addition rule: probability of drawing a heart OR a king
p_heart_or_king <- p_heart + p_king - p_king_of_hearts
p_heart_or_king

# Complement rule: probability of NOT drawing a heart
1 - p_heart
```

## 3. Discrete and Continuous Probability Distributions

- A **discrete distribution** describes outcomes that take on countable values (e.g. number of defective items).
- A **continuous distribution** describes outcomes that can take any value within a range (e.g. height, weight, time).

Every named distribution in R has four related functions with prefixes:
- `d` - density / probability mass function
- `p` - cumulative distribution function
- `q` - quantile function (inverse of `p`)
- `r` - random number generation


## 4. Normal, Binomial, Poisson, and Uniform Distributions

### Normal Distribution
Continuous, bell shaped, defined by mean and standard deviation. Used for naturally occurring measurements.

### Binomial Distribution
Discrete, models the number of successes in a fixed number of independent yes/no trials.

### Poisson Distribution
Discrete, models the number of events occurring in a fixed interval of time or space.

### Uniform Distribution
Every outcome in a range is equally likely (continuous or discrete).

```r
# ---- Normal distribution ----
dnorm(0, mean = 0, sd = 1)          # density at x = 0
pnorm(1.96, mean = 0, sd = 1)       # P(X <= 1.96)
qnorm(0.975, mean = 0, sd = 1)      # 97.5th percentile
rnorm(5, mean = 100, sd = 15)       # 5 random draws

# ---- Binomial distribution ----
dbinom(3, size = 10, prob = 0.5)    # P(X = 3 successes out of 10 trials)
pbinom(3, size = 10, prob = 0.5)    # P(X <= 3)
rbinom(5, size = 10, prob = 0.5)    # 5 random draws

# ---- Poisson distribution ----
dpois(2, lambda = 3)                # P(X = 2 events, average rate 3)
ppois(2, lambda = 3)                # P(X <= 2)
rpois(5, lambda = 3)                # 5 random draws

# ---- Uniform distribution ----
dunif(0.5, min = 0, max = 1)
punif(0.5, min = 0, max = 1)
runif(5, min = 0, max = 1)          # 5 random draws
```

## 5. Random Number Generation

R's random number functions are essential for simulation, bootstrapping, and Monte Carlo methods. Always set a seed with `set.seed()` to make your results reproducible.

```r
set.seed(42)                    # ensures reproducibility

random_normals <- rnorm(1000, mean = 50, sd = 10)
head(random_normals)
mean(random_normals)
sd(random_normals)

# Random sampling from a vector
sample(1:100, size = 10, replace = FALSE)
```

## 6. Distribution Visualization in R

Visualizing distributions helps you understand shape, spread, and central tendency at a glance.

```r
library(ggplot2)

set.seed(1)
normal_data <- data.frame(value = rnorm(5000, mean = 0, sd = 1))

ggplot(normal_data, aes(x = value)) +
  geom_histogram(aes(y = after_stat(density)), bins = 40, fill = "skyblue", color = "white") +
  stat_function(fun = dnorm, args = list(mean = 0, sd = 1), color = "red", linewidth = 1) +
  labs(title = "Simulated Standard Normal Distribution")

# Binomial distribution bar plot
binom_data <- data.frame(x = 0:10, prob = dbinom(0:10, size = 10, prob = 0.5))
ggplot(binom_data, aes(x = factor(x), y = prob)) +
  geom_col(fill = "orange") +
  labs(title = "Binomial Distribution (n = 10, p = 0.5)", x = "Number of Successes", y = "Probability")
```
