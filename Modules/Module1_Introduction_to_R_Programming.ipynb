## 1. Overview of R and RStudio

**R** is a free, open source programming language and software environment built specifically for statistical computing, data analysis, and graphics. It was created by Ross Ihaka and Robert Gentleman at the University of Auckland in the early 1990s and is now maintained by the R Development Core Team.

Why R is popular for data analysis:
- Purpose built for statistics, modelling, and visualization
- Huge ecosystem of packages on CRAN (Comprehensive R Archive Network)
- Strong data visualization tools, especially `ggplot2`
- Active community used heavily in academia, research, finance, and industry
- Free and open source, runs on Windows, macOS, and Linux

**RStudio** is the most widely used Integrated Development Environment (IDE) for R. It is not R itself - it is a separate application that makes R easier to work with. RStudio provides:
- A script editor with syntax highlighting and auto complete
- A console for running R code interactively
- An environment pane showing all variables currently in memory
- A plots pane for viewing charts
- A files/packages/help pane
- Built in project management and version control (Git) integration

## 3. R Syntax and Programming Fundamentals

R code is made up of expressions that are evaluated in the console or run from a script. Key syntax rules:
- Statements can be separated by a new line or a semicolon `;`
- Comments start with `#` and are ignored by R
- Assignment is usually done with `<-` (preferred) or `=`
- Function calls use parentheses: `function_name(arguments)`
- R is case sensitive: `x` and `X` are different objects


```r
# This is a comment - R ignores this line
x <- 10          # preferred assignment operator
y = 5             # also valid, less commonly used in R style guides
z <- x + y
print(z)

# Multiple statements on one line using a semicolon
a <- 2; b <- 3; print(a * b)

# Simple function call
sum(1, 2, 3)
```

## 4. Data Types, Variables, and Operators

### Core Data Types in R
| Type | Example | Description |
|------|---------|-------------|
| numeric | 3.14 | Decimal (double precision) numbers |
| integer | 5L | Whole numbers, marked with `L` |
| character | "hello" | Text strings |
| logical | TRUE / FALSE | Boolean values |
| complex | 2+3i | Complex numbers |

### Operators
- Arithmetic: `+  -  *  /  ^  %%  %/%`
- Relational: `==  !=  >  <  >=  <=`
- Logical: `&  |  !  &&  ||`
- Assignment: `<-  =  <<-`


```r
# Data type examples
num_val <- 3.14
int_val <- 5L
char_val <- "Data Science"
log_val <- TRUE

class(num_val)
class(int_val)
class(char_val)
class(log_val)

# Checking types
is.numeric(num_val)
is.character(char_val)

# Arithmetic operators
10 %% 3   # modulus - remainder of division
10 %/% 3  # integer division
2 ^ 4     # exponent

# Relational and logical operators
5 > 3
(5 > 3) & (2 < 1)
(5 > 3) | (2 < 1)
```

## 5. Working with Scripts

An R script is a plain text file with a `.R` extension containing a sequence of R commands. Scripts make your analysis reproducible and shareable.

Good practices when writing R scripts:
- Start with a header comment describing the purpose of the script
- Load all required packages at the top
- Break the script into clearly commented sections
- Use meaningful variable names
- Save scripts often and use version control where possible

You can run:
- A single line: place your cursor on the line and press `Ctrl+Enter` (Windows/Linux) or `Cmd+Enter` (macOS)
