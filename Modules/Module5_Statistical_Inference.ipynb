## 1. Sampling Techniques

Statistical inference lets us draw conclusions about a population using only a sample. Common sampling techniques include:
- **Simple random sampling**: every member has an equal chance of selection
- **Stratified sampling**: population divided into subgroups, then sampled within each
- **Systematic sampling**: selecting every k-th member from a list
- **Cluster sampling**: dividing the population into clusters, then randomly selecting whole clusters


```r
set.seed(1)
population <- 1:1000

# Simple random sampling
srs <- sample(population, size = 50)

# Systematic sampling - every 20th element
systematic <- population[seq(1, length(population), by = 20)]

# Stratified sampling using dplyr
library(dplyr)
df <- data.frame(id = population, group = rep(c("A", "B"), each = 500))
stratified <- df %>% group_by(group) %>% slice_sample(n = 25)
table(stratified$group)
```

## 2. Confidence Intervals

A confidence interval gives a range of plausible values for a population parameter, along with a confidence level (commonly 95 percent) expressing how confident we are that the interval contains the true value.

```r
set.seed(10)
sample_data <- rnorm(50, mean = 100, sd = 15)

# Manual 95 percent confidence interval for the mean
n <- length(sample_data)
xbar <- mean(sample_data)
se <- sd(sample_data) / sqrt(n)
margin <- qt(0.975, df = n - 1) * se
ci_lower <- xbar - margin
ci_upper <- xbar + margin
c(lower = ci_lower, upper = ci_upper)

# Using a built in t-test to get the same confidence interval directly
t.test(sample_data)$conf.int
```

## 3. Hypothesis Testing

Hypothesis testing is a formal procedure for deciding whether sample data supports a claim about a population.

- **Null hypothesis (H0)**: the default assumption, usually 'no effect' or 'no difference'
- **Alternative hypothesis (H1)**: the claim we are testing for
- **p-value**: probability of observing data this extreme if H0 is true
- **Significance level (alpha)**: commonly 0.05 - the threshold for rejecting H0

Decision rule: if p-value < alpha, reject the null hypothesis.

## 4. t-Tests

t-Tests compare means. There are three common types:
- **One sample t-test**: compares a sample mean to a known value
- **Two sample (independent) t-test**: compares means of two independent groups
- **Paired t-test**: compares means of two related measurements (before/after)


```r
set.seed(5)

# One sample t-test - is the sample mean different from 100?
sample1 <- rnorm(30, mean = 103, sd = 10)
t.test(sample1, mu = 100)

# Two sample independent t-test
group_a <- rnorm(30, mean = 50, sd = 5)
group_b <- rnorm(30, mean = 55, sd = 5)
t.test(group_a, group_b)

# Paired t-test - before and after measurements
before <- rnorm(20, mean = 70, sd = 8)
after <- before + rnorm(20, mean = 3, sd = 2)
t.test(before, after, paired = TRUE)
```

## 5. Chi-Square Tests

Chi-Square tests are used with categorical data.
- **Goodness of fit test**: does the observed distribution match an expected distribution?
- **Test of independence**: are two categorical variables related?


```r
# Goodness of fit test
observed <- c(18, 22, 20, 40)
expected_probs <- c(0.25, 0.25, 0.25, 0.25)
chisq.test(observed, p = expected_probs)

# Test of independence
survey_table <- matrix(c(30, 10, 20, 40), nrow = 2,
                        dimnames = list(Gender = c("Male", "Female"),
                                        Preference = c("Product A", "Product B")))
survey_table
chisq.test(survey_table)
```

## 6. ANOVA (Analysis of Variance)

ANOVA tests whether the means of three or more groups are significantly different from one another. A significant ANOVA result is usually followed by post-hoc tests (e.g. Tukey's HSD) to identify exactly which groups differ.

```r
set.seed(7)
scores <- c(rnorm(20, mean = 70, sd = 5),
            rnorm(20, mean = 75, sd = 5),
            rnorm(20, mean = 80, sd = 5))
group <- factor(rep(c("Method A", "Method B", "Method C"), each = 20))
anova_data <- data.frame(scores, group)

anova_model <- aov(scores ~ group, data = anova_data)
summary(anova_model)

# Post-hoc test to find which groups differ
TukeyHSD(anova_model)
```

## 7. Non-Parametric Statistical Tests

Non-parametric tests are used when data does not meet the assumptions required for t-tests or ANOVA (e.g. normality). Common alternatives:
- **Wilcoxon rank sum test**: non-parametric alternative to the independent t-test
- **Wilcoxon signed rank test**: non-parametric alternative to the paired t-test
- **Kruskal-Wallis test**: non-parametric alternative to one-way ANOVA


```r
set.seed(3)
group_x <- c(23, 25, 21, 30, 29, 100)   # includes an outlier
group_y <- c(20, 22, 24, 19, 21, 23)

# Wilcoxon rank sum test (Mann-Whitney U test)
wilcox.test(group_x, group_y)

# Kruskal-Wallis test - non-parametric alternative to ANOVA
kruskal.test(scores ~ group, data = anova_data)
```
